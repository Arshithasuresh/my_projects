```python
import pygame
```

```python
pygame.init()
```

Setup the game window

```python
w = 800
h = 600
screen = pygame.display.set_mode((w,h))
pygame.display.set_caption("Sample Game")
```

```python
#colors
red = (255,0,0)
blue = (0,0,255)
```

```python
#player properties
size = 50
player_x=(w-size)//2
player_y=(h-size)//2
player_speed = 5
```

```python
#game loop
running = True
while running:
  for event in pygame.event.get():
    if event.type==pygame.QUIT:
      running = False
    elif event.type == pygame.KEYDOWN:
      if event.key == pygame.K_LEFT:
        player_x -= player_speed
      elif event.key == pygame.K_RIGHT:
        player_x += player_speed
      elif event.key == pygame.K_UP:
        player_y -= player_speed
      elif event.key == pygame.K_DOWN:
        player_y += player_speed
  player_x = max(0,min(player_x,w-size))
  player_y = max(0,min(player_y,h-size))
  screen.fill(blue)
  pygame.draw.rect(screen,red,(player_x,player_y,size,size))
  pygame.display.flip()
pygame.quit()
```
